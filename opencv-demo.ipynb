```python
import cv2
```

```python
image = cv2.imread("C:\\Users\\Dell Latitude 7490\\Pictures\\road(1).jpg")

h, w = image.shape[:2]

print(f"Width: {w}, Height: {h}")
```

```python
(b, g, r) = image[100, 100]

print(f"Red: {r}, Green: {g}, Blue: {b}")

b = image[100,100,0]

print(f"b = {b}")
```

```python
roi = image[100 : 500, 200 : 700]
cv2.imshow("ROI", roi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
resized = cv2.resize(image, (200, 200))
cv2.imshow("Resized Image", resized)
cv2.waitKey(0)
```

```python
ratio = 200 / w

dim = (200, int(h * ratio))

resize_aspect = cv2.resize(image, dim)
cv2.imshow("Resized Image", resize_aspect)
cv2.waitKey(0)
```

```python
output = image.copy()

rectangle = cv2.rectangle(output, (100,100),(600, 600), (0, 255, 0), 5)

cv2.imshow("Rectangle", rectangle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
output = image.copy()

text = cv2.putText(output, 'OpenCV Demo', (200, 300),
                cv2.FONT_HERSHEY_SIMPLEX, 4, (153,102,204), 2)

cv2.imshow("Demo", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
